### **q2)** **What happens to agricultural production if cultivated area increases by 10%, assuming crop yield remains constant?**

```python
#Step 1 — Simulate a 10% increase in cultivated area

# Original cultivated area
area = df['Area'].to_numpy()

# Broadcasting: Increase every area by 10%
new_area = area * 1.10

# Calculate additional cultivated area
area_increase = new_area - area

print("Original Area:", area[:5])
print("New Area:", new_area[:5])
print("Additional Area:", area_increase[:5])
```

```python
# Step 2 — Estimate the impact on production

# Original production
production = df['Production'].to_numpy()

# Assume yield remains constant
new_production = production * 1.10

# Calculate additional production
production_increase = new_production - production

print("Original Production:", production[:5])
print("Estimated Production:", new_production[:5])
print("Additional Production:", production_increase[:5])
```

```python
# Step 3 — Compare different scenarios

# Simulate 5%, 10%, and 20% area increases
growth_rates = np.array([0.05, 0.10, 0.20])

# Broadcasting across all records and scenarios
projected_production = production[:, np.newaxis] * (1 + growth_rates)

# Calculate total production for each scenario
total_production = projected_production.sum(axis=0)

for rate, total in zip(growth_rates, total_production):
    print(f"Area Increase: {rate:.0%}")
    print(f"Estimated Total Production: {total:,.2f}")
    print("-----------------------------")
```

```python
# Step 4 — conclusion
print("========== ANALYTICAL CONCLUSION ==========")

for rate, total in zip(growth_rates, total_production):
    print(
        f"A {rate:.0%} increase in cultivated area "
        f"results in a {rate:.0%} increase in estimated production "
        f"under the constant-yield assumption."
    )

print("\nThis is a scenario analysis, not a prediction.")
print("Actual production depends on crop, rainfall, soil, and other factors.")
```

### **q3) Which agricultural records produce more output per hectare, and how can we identify inefficient production using NumPy without Python loops?**

```python
# Step 1 — Calculate production per hectare

# Extract NumPy arrays
area = df['Area'].to_numpy(dtype=float)
production = df['Production'].to_numpy(dtype=float)

# Vectorized calculation without loops
production_per_hectare = np.divide(
    production,
    area,
    out=np.full_like(production, np.nan),
    where=area > 0
)

# Add result to DataFrame
df['Production_Per_Hectare'] = production_per_hectare

display(df[
    ['State', 'Crop', 'Area', 'Production', 'Production_Per_Hectare']
].head(10))
```

```python
# Step 2 — Identify high-productivity records

# Calculate 90th percentile
threshold = np.nanpercentile(production_per_hectare, 90)

# Boolean indexing
high_productivity = df[
    df['Production_Per_Hectare'] > threshold
]

print("90th Percentile:", threshold)
print("High Productivity Records:", len(high_productivity))

display(
    high_productivity[
        ['State', 'Crop', 'Crop_Year', 'Production_Per_Hectare']
    ].sort_values(
        'Production_Per_Hectare',
        ascending=False
    ).head(10)
)
```

```python
# Step 3 — Compare productivity within the same crop

# Filter wheat
wheat = df[df['Crop'] == 'Wheat'].copy()

# Calculate productivity threshold for wheat
wheat_threshold = np.nanpercentile(
    wheat['Production_Per_Hectare'], 90
)

# Classify records using vectorization
wheat['Productivity_Level'] = np.where(
    wheat['Production_Per_Hectare'] >= wheat_threshold,
    'High Productivity',
    'Other'
)

display(
    wheat[
        ['State', 'Crop_Year', 'Production_Per_Hectare', 'Productivity_Level']
    ].head(10)
)
```

```python
# Step 4 — Generate the analytical conclusion

print("==========CONCLUSION ==========")

print("Total Records:", len(df))
print("High Productivity Records:", len(high_productivity))

print("\nOverall Median Production Per Hectare:",
      np.nanmedian(production_per_hectare))

print("\nWheat 90th Percentile Threshold:",
      wheat_threshold)

print("\nWheat Productivity Classification:")
print(wheat['Productivity_Level'].value_counts())

print("\nConclusion:")
print("1. NumPy vectorization efficiently calculates productivity across records.")
print("2. The 90th percentile identifies relatively high-productivity observations.")
print("3. Crop-specific comparisons are more meaningful than mixing different crops.")
print("4. High productivity does not automatically mean better profitability.")
```

### **q4) Are there inconsistent, missing, or suspicious agricultural records?**

```python


# Create a clean working copy
clean_df = df.copy()

# Standardize text columns
for col in ['Crop', 'Season', 'State']:
    clean_df[col] = clean_df[col].str.strip()

# Check missing values
print("Missing Values:")
print(clean_df.isnull().sum())

# Check duplicates
print("\nDuplicates:", clean_df.duplicated().sum())

# Check invalid negative values
numeric_cols = [
    'Area', 'Production', 'Annual_Rainfall',
    'Fertilizer', 'Pesticide', 'Yield'
]

print("\nNegative Values:")
print((clean_df[numeric_cols] < 0).sum())

# Check zero area and production
print("\nZero Area:", (clean_df['Area'] == 0).sum())
print("Zero Production:", (clean_df['Production'] == 0).sum())

# Check extreme area records
display(clean_df.nlargest(10, 'Area'))

# Check cleaned seasons
print("\nSeasons:", clean_df['Season'].unique())
```

```python
# Which states recorded the highest wheat yields between 2010 and 2020?

# Filter wheat records from 2010 to 2020
wheat_df = clean_df[
    (clean_df['Crop'] == 'Wheat') &
    (clean_df['Crop_Year'].between(2010, 2020))
].copy()

# Sort by yield
top_wheat = wheat_df.sort_values(
    by='Yield',
    ascending=False
)

display(
    top_wheat[
        ['State', 'Crop_Year', 'Area', 'Production', 'Yield']
    ].head(10)
)

# Count matching records
print("Total Wheat Records:", len(wheat_df))
```

```python
if not top_wheat.empty:
    best = top_wheat.iloc[0]

    print("========== CONCLUSION ==========")
    print("Highest Yield State:", best['State'])
    print("Year:", best['Crop_Year'])
    print("Yield:", best['Yield'])
else:
    print("No matching records found.")
```

```python
# Which Indian states have the highest average wheat yield?

# Group wheat records by state
state_wheat = (
    wheat_df.groupby('State')
    .agg(
        Avg_Yield=('Yield', 'mean'),
        Total_Production=('Production', 'sum'),
        Total_Area=('Area', 'sum'),
        Records=('Yield', 'count')
    )
    .reset_index()
    .sort_values('Avg_Yield', ascending=False)
)

display(state_wheat.head(10))
```

```python
if not state_wheat.empty:
    best_state = state_wheat.iloc[0]

    print("========== CONCLUSION ==========")
    print("Highest Average Wheat Yield:", best_state['State'])
    print("Average Yield:", round(best_state['Avg_Yield'], 2))
    print("Total Production:", best_state['Total_Production'])
    print("Records Analyzed:", best_state['Records'])
```

```python
# How can we combine wheat production and rainfall summaries for each state and year?

# Dataset 1: Production by state and year
production_df = (
    wheat_df.groupby(['State', 'Crop_Year'], as_index=False)
    .agg(Total_Production=('Production', 'sum'))
)

# Dataset 2: Rainfall by state and year
rainfall_df = (
    wheat_df.groupby(['State', 'Crop_Year'], as_index=False)
    .agg(Avg_Rainfall=('Annual_Rainfall', 'mean'))
)

# Merge using common columns
merged_df = pd.merge(
    production_df,
    rainfall_df,
    on=['State', 'Crop_Year'],
    how='inner',
    validate='one_to_one'
)

display(merged_df.head(10))

print("Merged Shape:", merged_df.shape)
```

```python
correlation = merged_df[
    'Avg_Rainfall'
].corr(merged_df['Total_Production'])

print("========== CONCLUSION ==========")
print("Rainfall-Production Correlation:", round(correlation, 4))

if pd.isna(correlation):
    print("Correlation could not be calculated.")
elif correlation > 0.3:
    print("Positive linear association.")
elif correlation < -0.3:
    print("Negative linear association.")
else:
    print("Weak linear association.")
```

```python
# How has wheat yield changed across Indian states from 2010 to 2020?

# Create state-wise, year-wise yield comparison
pivot_df = pd.pivot_table(
    wheat_df,
    values='Yield',
    index='State',
    columns='Crop_Year',
    aggfunc='mean'
)

display(pivot_df.round(2))
```

```python
# Compare states with observations in both years
growth_df = pivot_df[[2010, 2020]].dropna().copy()

# Exclude zero baseline yields
growth_df = growth_df[growth_df[2010] > 0]

# Percentage change
growth_df['Growth_Percent'] = (
    (growth_df[2020] - growth_df[2010]) /
    growth_df[2010]
) * 100

# Rank states
growth_df = growth_df.sort_values(
    'Growth_Percent',
    ascending=False
)

display(growth_df.head(10))
```

```python
print("========== FINAL CONCLUSION ==========")

if not growth_df.empty:
    best = growth_df.iloc[0]

    print("State With Highest Wheat Yield Growth:", growth_df.index[0])
    print("2010 Average Yield:", round(best[2010], 2))
    print("2020 Average Yield:", round(best[2020], 2))
    print("Growth Percentage:", round(best['Growth_Percent'], 2), "%")
    print("States Compared:", len(growth_df))
else:
    print("Insufficient comparable data for 2010 and 2020.")
```
